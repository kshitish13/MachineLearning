```python
import pandas as pd 

df=pd.read_csv('home_prices.csv')
df.head()
```

```python
df['color'].unique()
```

```python
df.corr()
```

```python
#error  bcoz there is a text column 
#first we need to convert it to numerical and we will use onehot encoding 

df_encoded=pd.get_dummies(data=df,columns=['color'],drop_first=True)
df_encoded.head()
```

```python
cm=df_encoded.corr()
cm
```

```python
cm_price=abs(cm['price_lakhs'])
cm_price
```

```python
cm_price[cm_price>0.2]
```

```python
cm_price[cm_price>0.2].index
```

```python
selected_features=cm_price[cm_price>0.2].index.drop('price_lakhs')
selected_features
```

```python
from sklearn.model_selection import train_test_split

X=df[selected_features]
y=df['price_lakhs']

X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=25) 
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error

model=LinearRegression()

model.fit(X_train,y_train)

y_pred=model.predict(X_test)
mse=mean_squared_error(y_test,y_pred)
r2=r2_score(y_test,y_pred)

print(f"R2 Score : {r2}, MSE : {mse}")
```

```python
from sklearn.neighbors import KNeighborsRegressor

model=KNeighborsRegressor()

model.fit(X_train,y_train)

y_pred=model.predict(X_test)
mse=mean_squared_error(y_test,y_pred)
r2=r2_score(y_test,y_pred)

print(f"R2 Score : {r2}, MSE : {mse}")
```
